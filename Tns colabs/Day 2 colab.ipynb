```python
# Creating a list
my_list = [10,20,30,40,20,"Apple",True,1.5,"Kartik"]
print("List:",my_list)
print(type(my_list))
```

```python
# append() → Adds a single element to the end of the list
numbers = [15, 22, 30]
numbers.append(29)
numbers.append(52)
print("After append:", numbers)
```

```python
# extend() → Adds all elements from another iterable (like another list)
numbers = [1, 2, 3]
print (numbers)
letter = ['A','B','C'] #tuple
print(type(letter))
numbers.extend(letter)
print("After extend:", numbers)
print(type(numbers))
letter.extend(numbers)
print(letter)
```

```python
# insert() → Inserts an element at a specified index
numbers = [1, 3, 4,5]
numbers.insert(3,"wetgdyhrfd")
print("After insert:", numbers)
```

```python
# remove() → Removes the first occurrence of a specified value
numbers = [1,3,5,7,6,3,7,2,4,9,5,8,4,7,9,"Apple"]
numbers.remove(5)
print("after remove:", numbers)
```

```python
# pop() → Removes and returns element at given index (default is last)
numbers = [10,20,30,40,50]
popped_value = numbers.pop(4)
print("Popped valued:",popped_value)
print("After pop:", numbers)
```

```python
# clear() → Removes all elements from the list
numbers = [10,20,30]
numbers.clear()
print("After clear", numbers)
```

```python
# index() → Returns the index of the first occurrence of a specified value
numbers = [10,10,20,20,30,40,20]
index_pop= numbers.index(20)
print("Index of 20:",index_pop)
```

```python
# count() → Counts the number of times a value occurs in the list
numbers = [1,2,4,6,7,4,8,9,4,8,4,6,0,8]
count_numbers = numbers.count(4)
print("COunt of the value:",count_numbers)
```

```python
# reverse() → Reverses the list in-place
numbers = [1,2,4,5,6,7,"Piyush"]
numbers.reverse()
print("After reverse:",numbers)
```

```python
# sort() → Sorts the list in ascending order
numbers = ['d','g','k']
numbers.sort()
print("After sort:",numbers)
```

```python
# copy() → Returns a shallow copy of the list
numbers = [1, 2, 3,"Rohit"]
copy_numbers = numbers.copy()
print("Copied list:", copy_numbers)
```

```python
# len() → Returns the number of elements in the list
numbers = [1,2,4,6,3,6,8,3]
result = len(numbers)
print("lenght of list:",result)
```

```python
# sum() → Returns the sum of all numeric elements in the list
numbers = [1,2,3,4,5,6,7,8]
total = sum(numbers)
print("sum of list:",total)
```

```python
# max() → Returns the largest element in the list
# min() → Returns the smallest element in the list
numbers = [2,345,7,223,76,87,43,68,43,88]
print("Max value:", max(numbers))
print("Min value:", min(numbers))
```

Problem:
A. You are given a list of integers. Write a Python program to perform the following operations sequentially:
1.Append the number 100 to the end of the list.

2.Insert the number 50 at the 3rd position (index 2).

3.Remove the first occurrence of the number 20.

4.Count how many times 10 appears in the list.

5.Reverse the list.

6.Sort the list in ascending order.

7.Find the maximum and minimum values in the list.

8.Calculate the sum of all elements in the list.

```python
l1 = [12,45,36,10,65,20,10,43,57,35,65,89,32,10]
l1.append(100)
print("After apped 100:",l1)
l1.insert(2,50)
print("After insert 50:",l1)
l1.remove(20)
print("After remove 20:",l1)
print("count of 10:",l1.count(10))
l1.reverse()
print("After reverse:",l1)
l1.sort()
print("After sort:",l1)
max(l1)
print("Max value:",max(l1))
min(l1)
print("Min value:",min(l1))
sum(l1)
print("Sum of list:",sum(l1))
```

```python
person = {"name": "Alex", "age": 25, "city": "Mumbai", "Country":"India","height":165}

# Remove last inserted item
last_item = person.popitem()
print("Removed last item:", last_item)
print(person)
last_item = person.popitem()
print("Removed last item:", last_item)
print(person)
```

```python
person = {"age":2345}

# Get value if exists, else set a default
value = person.setdefault("country", "India")
print(value)
print(person)
```
